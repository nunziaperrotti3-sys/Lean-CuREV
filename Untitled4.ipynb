```python
import re

# Funzione pratica: pulisce e valuta se un commento ha senso senza usare AI
def filtro_rapido(testo):
    if not isinstance(testo, str):
        return False
    testo_pulito = testo.strip()
    # Regola 1: Scarta commenti con meno di 4 parole (es. "Same here", "LGTM", "Why?")
    parole = testo_pulito.split()
    if len(parole) < 4:
        return False
    # Regola 2: Scarta commenti che contengono solo un link URL
    if re.match(r'^https?://\S+$', testo_pulito):
        return False
    return True

# Applichiamo il filtro alle 500 righe
campione['passa_filtro_rapido'] = campione['comment'].apply(filtro_rapido)

scartati = campione[~campione['passa_filtro_rapido']]
tenuti = campione[campione['passa_filtro_rapido']]

print(f"Totale commenti analizzati: {len(campione)}")
print(f"Scartati in 0.01 secondi senza AI: {len(scartati)} ({len(scartati)/len(campione)*100:.1f}%)")
print(f"Tenuti per la fase successiva: {len(tenuti)}")
print("\nEsempi di commenti scartati:")
for c in scartati['comment'].head(5):
    print("-", repr(c))
```

```python
!pip install -q transformers torch

from transformers import pipeline
import time

start_time = time.time()

# 1. FASE 1 AFFINATA: Lista nera di convenevoli inutili (salva i commenti tecnici brevi!)
frasi_inutili = {"lgtm", "thanks", "thank you", "done", "fixed", "+1", "same here", "nice", "ok", "okay"}

def filtro_intelligente(testo):
    if not isinstance(testo, str):
        return False
    pulito = testo.strip().lower().rstrip(".!")
    if len(pulito) < 3 or pulito in frasi_inutili:
        return False
    return True

campione['fase1_ok'] = campione['comment'].apply(filtro_intelligente)
validi = campione[campione['fase1_ok']].copy()

# 2. FASE 2 LEGGERA: Micro-modello (DistilBERT) per individuare tono negativo/rude da riscrivere
classificatore_tono = pipeline("sentiment-analysis", model="distilbert-base-uncased-finetuned-sst-2-english", truncation=True)

# Analizziamo i commenti validi
risultati = classificatore_tono(validi['comment'].tolist(), batch_size=32)
validi['tono'] = [r['label'] for r in risultati]
validi['confidenza'] = [r['score'] for r in risultati]

# Regola pratica: riscriviamo SOLO se il tono è molto negativo (confidenza > 0.90)
da_riscrivere = validi[(validi['tono'] == 'NEGATIVE') & (validi['confidenza'] > 0.90)]
gia_buoni = validi.drop(da_riscrivere.index)

tempo_totale = time.time() - start_time

print(f"⏱️ Tempo totale elaborazione (500 righe): {tempo_totale:.2f} secondi")
print(f"✅ Commenti già buoni (lasciati originali): {len(gia_buoni)} ({len(gia_buoni)/len(campione)*100:.1f}%)")
print(f"✍️ Commenti da riscrivere con LLM piccolo: {len(da_riscrivere)} ({len(da_riscrivere)/len(campione)*100:.1f}%)")
print("\nEsempi intercettati come 'da riscrivere':")
for c in da_riscrivere['comment'].head(3):
    print("-", repr(c))
```

```python
!pip install -q datasets transformers torch pandas

import time
import pandas as pd
from datasets import load_dataset
from transformers import pipeline

start_time = time.time()

# 1. Carica il campione di 500 righe
ds = load_dataset("OussamaBS/CuREV-plus", split="train", streaming=True)
campione = pd.DataFrame(list(ds.take(500)))

# 2. FASE 1: Filtro regole anti-chiacchiere
frasi_inutili = {"lgtm", "thanks", "thank you", "done", "fixed", "+1", "same here", "nice", "ok", "okay"}
validi = campione[campione['comment'].apply(lambda t: isinstance(t, str) and len(t.strip()) >= 3 and t.strip().lower().rstrip(".!") not in frasi_inutili)].copy()

# 3. FASE 2: Classificatore specifico di TOSSICITÀ su GPU (device=0)
classificatore_tossicita = pipeline("text-classification", model="unitary/toxic-bert", device=0, truncation=True)

risultati = classificatore_tossicita(validi['comment'].tolist(), batch_size=64)
validi['score_tossicita'] = [r['score'] for r in risultati]

# Regola pratica: riscriviamo solo se tossico (> 0.50) o troppo prolisso (> 25 parole)
da_riscrivere = validi[(validi['score_tossicita'] > 0.50) | (validi['comment'].str.split().str.len() > 25)]
gia_buoni = validi.drop(da_riscrivere.index)

tempo_totale = time.time() - start_time

print(f"⚡ Tempo totale con GPU: {tempo_totale:.2f} secondi")
print(f"✅ Commenti già buoni (lasciati originali): {len(gia_buoni)} ({len(gia_buoni)/len(campione)*100:.1f}%)")
print(f"✍️ Commenti selezionati per la riscrittura: {len(da_riscrivere)} ({len(da_riscrivere)/len(campione)*100:.1f}%)")
```

```python
import torch
from transformers import AutoModelForCausalLM, AutoTokenizer

start_llm = time.time()

# Carichiamo un LLM piccolo, moderno e veloce (3 miliardi di parametri, sta comodamente nella GPU T4 gratuita!)
modello_id = "Qwen/Qwen2.5-3B-Instruct"
tokenizer = AutoTokenizer.from_pretrained(modello_id)
model = AutoModelForCausalLM.from_pretrained(modello_id, torch_dtype=torch.float16, device_map="auto")

def riscrivi_pratico(commento):
    # Prompt diretto ed essenziale: niente 10 esempi, niente spiegazioni prolisse
    messaggi = [
        {"role": "system", "content": "You are a helpful code reviewer. Rewrite the given code review comment to be polite, clear, and concise (max 2 sentences). Keep the exact technical meaning. Output ONLY the rewritten comment, nothing else."},
        {"role": "user", "content": commento}
    ]
    testo_input = tokenizer.apply_chat_template(messaggi, tokenize=False, add_generation_prompt=True)
    inputs = tokenizer([testo_input], return_tensors="pt").to("cuda")

    outputs = model.generate(**inputs, max_new_tokens=60, temperature=0.3, do_sample=True)
    risposta = tokenizer.decode(outputs[0][inputs.input_ids.shape[1]:], skip_special_tokens=True)
    return risposta.strip()

# Testiamo su un campione dei primi 10 commenti selezionati da riscrivere (ci mette ~5 secondi!)
sottoinsieme_test = da_riscrivere.head(10).copy()
sottoinsieme_test['nostra_riscrittura'] = sottoinsieme_test['comment'].apply(riscrivi_pratico)

tempo_llm = time.time() - start_llm
print(f"🚀 Tempo riscrittura 10 commenti con Qwen-3B: {tempo_llm:.2f} secondi (media {tempo_llm/10:.2f}s a commento)\n")

# Stampiamo 3 confronti diretti: Originale vs Autori (70B) vs Nostro Metodo Pratico (3B)
for idx, row in sottoinsieme_test.head(3).iterrows():
    print("="*60)
    print(f"🔴 ORIGINALE:        {row['comment']}")
    print(f"🔵 AUTORI (CuREV+):  {row['reformulated_comment']}")
    print(f"🟢 NOSTRO (Lean-3B): {row['nostra_riscrittura']}")
```
